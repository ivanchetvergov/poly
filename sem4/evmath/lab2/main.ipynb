```python
import numpy as np

x = np.array([-0.1, 0.3, 0.5, 0.75, 0.9, 1.1, 1.3, 1.5, 1.7, 1.9,
              2.1, 2.3, 2.5, 2.7, 2.9, 3.1, 3.3, 3.5, 3.7, 3.9])
y = np.array([-0.08, -0.14, -0.89, 0.36, 0.44, 0.48, 0.27, 0.39, 0.50, 0.48,
              0.69, 0.50, 0.31, 0.37, 0.43, 0.33, 0.31, 0.09, 0.08, 0.03])

n = len(x)
sx = np.sum(x)
sx2 = np.sum(x**2)
sy = np.sum(y)
sxy = np.sum(x * y)

print(f"n = {n}")
print(f"sum x = {sx:.4f}")
print(f"sum x2 = {sx2:.4f}")
print(f"sum y = {sy:.4f}")
print(f"sum xy = {sxy:.4f}")
```

```python
# Нормальная система: A * a = b
A = np.array([[n, sx],
              [sx, sx2]])
b = np.array([sy, sxy])

a0_1, a1_1 = np.linalg.solve(A, b)

# Аппроксимирующий полином и отклонение
P1 = a0_1 + a1_1 * x
delta1 = np.sqrt(np.sum((y - P1)**2) / n)

print(f"a0 = {a0_1:.4f}, a1 = {a1_1:.4f}")
print(f"P1(x) = {a0_1:.4f} + ({a1_1:.4f})*x")
print(f"delta2(P1) = {delta1:.4f}")
```

```python
sx3 = np.sum(x**3)
sx4 = np.sum(x**4)
sx2y = np.sum(x**2 * y)

# Нормальная система: A * a = b
A2 = np.array([[n, sx, sx2],
               [sx, sx2, sx3],
               [sx2, sx3, sx4]])
b2 = np.array([sy, sxy, sx2y])

a0_2, a1_2, a2_2 = np.linalg.solve(A2, b2)

# Аппроксимирующий полином и отклонение
P2 = a0_2 + a1_2 * x + a2_2 * x**2
delta2 = np.sqrt(np.sum((y - P2)**2) / n)

print(f"a0 = {a0_2:.4f}, a1 = {a1_2:.4f}, a2 = {a2_2:.4f}")
print(f"P2(x) = {a0_2:.4f} + ({a1_2:.4f})*x + ({a2_2:.4f})*x^2")
print(f"delta2(P2) = {delta2:.4f}")
```

```python
import matplotlib.pyplot as plt

# Для графика возьмем сетку от -0.2 до 4.0
x_fine = np.linspace(-0.2, 4.0, 300)

# Значения полиномов на мелкой сетке
P1_fine = a0_1 + a1_1 * x_fine
P2_fine = a0_2 + a1_2 * x_fine + a2_2 * x_fine**2

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', zorder=5, label='Исходные данные')
plt.plot(x_fine, P1_fine, color='black', linewidth=1.5, label=r'$P_1(x)$ — линейная')
plt.plot(x_fine, P2_fine, color='red', linewidth=1.5, label=r'$P_2(x)$ — квадратичная')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Аппроксимация методом наименьших квадратов')
plt.legend()
plt.grid(True)
plt.tight_layout()
# Сохраняем график в папку отчета
plt.savefig('report/pct/approximation.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
import os
```

```python
X = np.array([11.153, 11.454, 11.673, 11.879, 12.009, 12.231, 12.549]).reshape(-1, 1)
y = np.array([-3.234, 5.321, -1.123, 0.393, 8.939, 141.231, 15.001])
```

```python
class PolynomialRegressor:
    def __init__(self, degree: int):
        self.degree = degree
        self.weights = None

    def fit(self, X: np.ndarray, y: np.ndarray):
        if X.ndim == 1:
            X = X.reshape(-1, 1)
        if y.ndim == 1:
            y = y.reshape(-1, 1)

        # создание матрицы признаков для полинома степени degree
        X_poly = np.ones((X.shape[0], self.degree + 1))
        for d in range(1, self.degree + 1):
            X_poly[:, d] = X.flatten() ** d

        # решение уравнения: w = (X^T X)^-1 X^T y
        XTX = X_poly.T @ X_poly
        XTy = X_poly.T @ y
        self.weights = np.linalg.inv(XTX) @ XTy

        return self

    def predict(self, X: np.ndarray) -> np.ndarray:
        if X.ndim == 1:
            X = X.reshape(-1, 1)

        X_poly = np.ones((X.shape[0], self.degree + 1))
        for d in range(1, self.degree + 1):
            X_poly[:, d] = X.flatten() ** d

        return (X_poly @ self.weights).flatten()

    def rmse(self, X: np.ndarray, y: np.ndarray) -> float:
        y_pred = self.predict(X)
        return np.sqrt(mean_squared_error(y, y_pred))

    def get_coefficients(self) -> np.ndarray:

        return self.weights.flatten()


lin_reg = PolynomialRegressor(degree=1)
lin_reg.fit(X, y)
lin_rmse = lin_reg.rmse(X, y)
lin_coeffs = lin_reg.get_coefficients()

print("--- Linear Model ---")
print(f"y = {lin_coeffs[1]:.4f}x + {lin_coeffs[0]:.4f}")
print(f"RMSE: {lin_rmse:.4f}")

quad_reg = PolynomialRegressor(degree=2)
quad_reg.fit(X, y)
quad_rmse = quad_reg.rmse(X, y)
quad_coeffs = quad_reg.get_coefficients()

print("\n--- Quadratic Model ---")
print(f"y = {quad_coeffs[2]:.4f}x^2 + {quad_coeffs[1]:.4f}x + {quad_coeffs[0]:.4f}")
print(f"RMSE: {quad_rmse:.4f}")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='black', label='Original Data', zorder=5)

x_range = np.linspace(X.min() - 0.1, X.max() + 0.1, 200).reshape(-1, 1)
y_range_lin = lin_reg.predict(x_range)
plt.plot(x_range, y_range_lin, color='blue', linestyle='--', label=f'Linear (RMSE: {lin_rmse:.2f})')

y_range_quad = quad_reg.predict(x_range)
plt.plot(x_range, y_range_quad, color='red', label=f'Quadratic (RMSE: {quad_rmse:.2f})')

plt.title('Linear vs Quadratic Regression')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3);
```

```python
model_lin = LinearRegression().fit(X, y)
y_pred_lin = model_lin.predict(X)
rmse_lin = np.sqrt(mean_squared_error(y, y_pred_lin))
```

```python
poly = PolynomialFeatures(degree=2)
xs_poly = poly.fit_transform(X)
model_quad = LinearRegression().fit(xs_poly, y)
y_pred_quad = model_quad.predict(xs_poly)
rmse_quad = np.sqrt(mean_squared_error(y, y_pred_quad))
```

```python
print("--- Linear ---")
print(f"y = {model_lin.coef_[0]:.4f}x + {model_lin.intercept_:.4f}")
print(f"RMSE: {rmse_lin:.4f}")

print("\n--- Quadratic ---")
print(f"y = {model_quad.coef_[2]:.4f}x^2 + {model_quad.coef_[1]:.4f}x + {model_quad.intercept_:.4f}")
print(f"RMSE: {rmse_quad:.4f}")
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='black', label='Original Data', zorder=5)

x_range = np.linspace(X.min() - 0.1, X.max() + 0.1, 200).reshape(-1, 1)
y_range_lin = model_lin.predict(x_range)
plt.plot(x_range, y_range_lin, color='blue', linestyle='--', label=f'Linear (RMSE: {rmse_lin:.2f})')

x_range_poly = poly.transform(x_range)
y_range_quad = model_quad.predict(x_range_poly)
plt.plot(x_range, y_range_quad, color='red', label=f'Quadratic (RMSE: {rmse_quad:.2f})')

plt.title('Linear vs Quadratic Regression')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True, alpha=0.3);
```
